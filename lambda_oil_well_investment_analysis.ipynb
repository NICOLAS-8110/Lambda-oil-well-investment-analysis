```python

df_1_features = df_1.drop(['product', 'id'], axis=1)
df_1_target = df_1['product']

df_2_features = df_2.drop(['product', 'id'], axis=1)
df_2_target = df_2['product']

df_3_features = df_3.drop(['product', 'id'], axis=1)
df_3_target = df_3['product']

df_1_features_train, df_1_features_valid, df_1_target_train, df_1_target_valid = train_test_split(df_1_features, df_1_target, test_size=0.25, random_state=12345)
df_2_features_train, df_2_features_valid, df_2_target_train, df_2_target_valid = train_test_split(df_2_features, df_2_target, test_size=0.25, random_state=12345)
df_3_features_train, df_3_features_valid, df_3_target_train, df_3_target_valid = train_test_split(df_3_features, df_3_target, test_size=0.25, random_state=12345)
```

```python
def model_train(features_train, target_train, features_valid, target_valid):
    model = LinearRegression()
    model.fit(features_train, target_train)
    prediction = model.predict(features_valid)
    return prediction, target_valid
```

```python
prediction_1, target_valid_1 = model_train(df_1_features_train,
                                           df_1_target_train,
                                           df_1_features_valid,
                                           df_1_target_valid)

prediction_2, target_valid_2 = model_train(df_2_features_train,
                                           df_2_target_train,
                                           df_2_features_valid,
                                           df_2_target_valid)

prediction_3, target_valid_3 = model_train(df_3_features_train,
                                           df_3_target_train,
                                           df_3_features_valid,
                                           df_3_target_valid)
```

```python

MSE_1 = mean_squared_error(target_valid_1, prediction_1)
RMSE_1 = np.sqrt(MSE_1)
mean_RMSE_1 = prediction_1.mean()
print('El RMSE de la primera región es:', RMSE_1)
print('El volumen medio de reservas predicho de la región 1 es:', mean_RMSE_1)
```

```python
inversion = 100000000
pozos = 200
precio_por_unidad = 4500
```

```python
promedio_por_pozo = inversion / pozos
volumen_minimo_rentable = promedio_por_pozo / precio_por_unidad
print('La inversión por pozo es:', promedio_por_pozo)
print('El volumen minimo rentable por pozo es:', volumen_minimo_rentable)
```

El hecho de que el promedio general de reservas sea bajo no significa que el proyecto sea inviable. El promedio mezcla pozos muy productivos con otros poco rentables, y lo que realmente importa es la selección de los 200 mejores pozos de cada región. Eso abre una hipótesis razonable: esos 200 pueden tener un promedio mucho más alto que el global. Sin embargo, la viabilidad final todavía no está confirmada, se comprobará en los siguientes pasos, cuando se calculen las ganancias con esos pozos y se evalúe el riesgo mediante bootstrapping.

```python
def ganancias(predicciones, target_valid):
    top_200 = predicciones.sort_values(ascending=False)
    top_200 = top_200[:200]
    indices_top_200 = top_200.index
    valores_reales_top_200 = target_valid.loc[indices_top_200]
    volumen_total_real = valores_reales_top_200.sum()
    ganancia = (volumen_total_real * precio_por_unidad) - inversion
    return ganancia
```

```python
def evaluar_region_bootstrapping(prediction, target_valid, nombre_region, n_muestras=1000, n_puntos=500, state=None):
    if state is None:
        state = np.random.RandomState(12345)
        
    resultados = []
    prediction_series = pd.Series(prediction, index=target_valid.index)
    
    for i in range(n_muestras):
        muestra = prediction_series.sample(n=n_puntos, replace=True, random_state=state)
        indices_muestra = muestra.index
        valores_reales = target_valid.loc[indices_muestra]
        
        ganancia_sample = ganancias(muestra, valores_reales)
        resultados.append(ganancia_sample)
        
    resultados = pd.Series(resultados)
    
    ganancia_promedio = resultados.mean()
    limite_inferior = resultados.quantile(0.025)
    limite_superior = resultados.quantile(0.975)
    riesgo_perdida = (resultados < 0).mean() * 100
    
    print(f"--- {nombre_region} ---")
    print(f"Ganancia promedio: {ganancia_promedio:,.2f} USD")
    print(f"Intervalo de confianza 95%: [{limite_inferior:,.2f}, {limite_superior:,.2f}] USD")
    print(f"Riesgo de pérdida: {riesgo_perdida:.1f} %\n")
    
    return resultados, ganancia_promedio, (limite_inferior, limite_superior), riesgo_perdida
```

```python
resultados_1, mean_1, ic_1, risk_1 = evaluar_region_bootstrapping(
    prediction_1, target_valid_1, "Región 1"
)

resultados_2, mean_2, ic_2, risk_2 = evaluar_region_bootstrapping(
    prediction_2, target_valid_2, "Región 2"
)

resultados_3, mean_3, ic_3, risk_3 = evaluar_region_bootstrapping(
    prediction_3, target_valid_3, "Región 3"
)
```

A partir de la evaluación de los modelos de regresión lineal y de las simulaciones con la técnica de bootstrapping con 1000 muestras de 500 puntos para seleccionar los 200 mejores pozos, se recomienda de manera categórica la selección de la Región 2 (geo_data_1.csv) para la inversión y apertura de los nuevos pozos petroleros. La Región 2 generó el mayor beneficio promedio con 6,652,410.58 USD, superando significativamente a la Región 1 (6,007,352.44 USD) y a la Región 3 (6,155,597.23 USD). Además, presentó el riesgo de pérdida más bajo de todas las opciones evaluadas, alcanzando apenas un 0.3%, valor ampliamente inferior al umbral máximo permitido por el negocio (2.5%). En contraste, la Región 3 queda descartada por superar dicho límite con un 3.0% de riesgo, mientras que la Región 1 presentó un riesgo del 2.0%. Aunque el promedio general de reservas de la Región 2 era el más bajo (68.73 unidades vs. las 111.11 requeridas para el punto de equilibrio individual), la excepcional precisión de su modelo (RMSE de 0.89) permitió predecir e identificar con exactitud los 200 pozos verdaderamente más rentables dentro de cada muestra. Esto se refleja también en un intervalo de confianza al 95% holgadamente positivo (1,579,884.81 USD, 11,976,415.87 USD), garantizando una inversión segura, altamente rentable y alineada con los criterios corporativos.
